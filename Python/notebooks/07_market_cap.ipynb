```python
import pandas as pd
from pathlib import Path

FILE_PATH = Path(
    r"C:\Users\KIIT\OneDrive\Desktop\Financial Analytics Project\Raw Data\market_cap.xlsx"
)

raw = pd.read_excel(
    FILE_PATH,
    sheet_name="Sheet1"
)

print("Shape:", raw.shape)

print("\nColumns:")
print(raw.columns.tolist())

print("\nData types:")
print(raw.dtypes)

print("\nMissing values:")
print(raw.isna().sum())

print("\nDuplicate IDs:")
print(raw["id"].duplicated().sum())

print(
    "\nDuplicate company-year:",
    raw.duplicated(
        subset=["company_id", "year"]
    ).sum()
)

print("\nYear range:")
print(raw["year"].min(), "to", raw["year"].max())
```

```python
raw.describe().T
```

```python
checks = {
    "Market cap < 0": (raw["market_cap_crore"] < 0).sum(),
    "Enterprise value < 0": (raw["enterprise_value_crore"] < 0).sum(),
    "P/E < 0": (raw["pe_ratio"] < 0).sum(),
    "P/B < 0": (raw["pb_ratio"] < 0).sum(),
    "EV/EBITDA < 0": (raw["ev_ebitda"] < 0).sum(),
    "Dividend yield < 0": (raw["dividend_yield_pct"] < 0).sum(),
}

for check, count in checks.items():
    print(f"{check}: {count}")
```

```python
from pathlib import Path

CLEANED_DATA = Path(
    r"C:\Users\KIIT\OneDrive\Desktop\Financial Analytics Project\Cleaned Data\valuation"
)

CLEANED_DATA.mkdir(parents=True, exist_ok=True)

valuation_clean = raw.copy()

valuation_clean.to_csv(
    CLEANED_DATA / "valuation_clean.csv",
    index=False
)

print("Saved:", CLEANED_DATA / "valuation_clean.csv")
```
