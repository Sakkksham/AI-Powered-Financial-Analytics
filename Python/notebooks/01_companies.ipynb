```python
print("Face value <= 0:",
      (companies["face_value"] <= 0).sum())

print("Book value <= 0:",
      (companies["book_value"] <= 0).sum())

print("ROCE < 0:",
      (companies["roce_percentage"] < 0).sum())

print("ROE < 0:",
      (companies["roe_percentage"] < 0).sum())
```

```python
companies_clean = companies.copy()
```

```python
companies_clean.shape
```

```python
quality_report = pd.DataFrame({
    "missing_count": companies_clean.isna().sum(),
    "missing_pct": companies_clean.isna().mean() * 100,
    "dtype": companies_clean.dtypes.astype(str)
})

quality_report.sort_values(
    "missing_count",
    ascending=False
)
```

```python
print("=" * 60)
print("COMPANY MASTER DATASET - FINAL VALIDATION")
print("=" * 60)

print(f"Rows: {companies_clean.shape[0]}")
print(f"Columns: {companies_clean.shape[1]}")

print("\nDuplicate IDs:",
      companies_clean["id"].duplicated().sum())

print("Duplicate company names:",
      companies_clean["company_name"].duplicated().sum())

print("\nMissing values:")
print(
    companies_clean.isna()
    .sum()
    .sort_values(ascending=False)
)

print("\nFinancial validation:")

print(
    "Face value <= 0:",
    (companies_clean["face_value"] <= 0).sum()
)

print(
    "Book value <= 0:",
    (companies_clean["book_value"] <= 0).sum()
)

print(
    "ROCE < 0:",
    (companies_clean["roce_percentage"] < 0).sum()
)

print(
    "ROE < 0:",
    (companies_clean["roe_percentage"] < 0).sum()
)

print("\nFinal shape:", companies_clean.shape)
```

```python
from pathlib import Path

CLEANED_DATA = Path(
    r"C:\Users\KIIT\OneDrive\Desktop\Financial Analytics Project\Cleaned Data"
)

CLEANED_DATA.mkdir(parents=True, exist_ok=True)

companies_clean.to_csv(
    CLEANED_DATA / "companies_clean.csv",
    index=False
)

print("Saved successfully:")
print(CLEANED_DATA / "companies_clean.csv")
```
