```python
import pandas as pd
from pathlib import Path

RAW_DATA = Path(
    r"C:\Users\KIIT\OneDrive\Desktop\Financial Analytics Project\Raw Data"
)

file_path = RAW_DATA / "profitandloss.xlsx"

raw = pd.read_excel(
    file_path,
    sheet_name="Profit & Loss",
    header=1
)

print("Rows before:", len(raw))

# Remove exact duplicate financial records while ignoring ID
dedup_cols = [c for c in raw.columns if c != "id"]

profit_loss_clean = (
    raw
    .drop_duplicates(subset=dedup_cols, keep="first")
    .copy()
)

print("Rows after:", len(profit_loss_clean))
print("Rows removed:", len(raw) - len(profit_loss_clean))
```

```python
duplicate_company_year = (
    profit_loss_clean
    .duplicated(["company_id", "year"])
    .sum()
)

print("Duplicate company-year:", duplicate_company_year)
print("Duplicate IDs:", profit_loss_clean["id"].duplicated().sum())
```

```python
print("Missing values:")
print(
    profit_loss_clean.isna().sum()
)
```

```python
print("Negative sales:",
      (profit_loss_clean["sales"] < 0).sum())

print("Negative expenses:",
      (profit_loss_clean["expenses"] < 0).sum())

print("Negative EPS:",
      (profit_loss_clean["eps"] < 0).sum())

print("Tax % below 0:",
      (profit_loss_clean["tax_percentage"] < 0).sum())

print("Tax % above 100:",
      (profit_loss_clean["tax_percentage"] > 100).sum())

print("Dividend payout below 0:",
      (profit_loss_clean["dividend_payout"] < 0).sum())
```
