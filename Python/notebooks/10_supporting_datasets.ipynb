```python
from pathlib import Path
import pandas as pd

STOCK_FILE = Path(
    r"C:\Users\KIIT\OneDrive\Desktop\Financial Analytics Project\Raw Data\stock_prices.xlsx"
)

stock_prices = pd.read_excel(
    STOCK_FILE,
    sheet_name="Sheet1"
)

# Standardize date
stock_prices["date"] = pd.to_datetime(
    stock_prices["date"],
    errors="coerce"
)

# Remove exact duplicates if any
before = len(stock_prices)

stock_prices = stock_prices.drop_duplicates(
    subset=[
        "company_id",
        "date",
        "open_price",
        "high_price",
        "low_price",
        "close_price",
        "volume",
        "adjusted_close"
    ]
).copy()

after = len(stock_prices)

print("Rows before:", before)
print("Rows after:", after)
print("Rows removed:", before - after)

# Validation
print("\nMissing values:")
print(stock_prices.isna().sum())

print("\nDuplicate company-date:",
      stock_prices.duplicated(
          subset=["company_id", "date"]
      ).sum())

print("\nNegative values:")
print(
    (stock_prices[
        [
            "open_price",
            "high_price",
            "low_price",
            "close_price",
            "volume",
            "adjusted_close"
        ]
    ] < 0).sum()
)

# OHLC consistency checks
stock_prices["ohlc_invalid"] = (
    (stock_prices["high_price"] <
     stock_prices[
         ["open_price", "close_price", "low_price"]
     ].max(axis=1))
    |
    (stock_prices["low_price"] >
     stock_prices[
         ["open_price", "close_price", "high_price"]
     ].min(axis=1))
)

print("\nOHLC consistency flags:",
      stock_prices["ohlc_invalid"].sum())
```

```python
stock_prices_clean = stock_prices[
    [
        "id",
        "company_id",
        "date",
        "open_price",
        "high_price",
        "low_price",
        "close_price",
        "volume",
        "adjusted_close"
    ]
].copy()

STOCK_CLEANED = Path(
    r"C:\Users\KIIT\OneDrive\Desktop\Financial Analytics Project\Cleaned Data\stock_prices"
)

STOCK_CLEANED.mkdir(parents=True, exist_ok=True)

output_path = STOCK_CLEANED / "stock_prices_clean.csv"

stock_prices_clean.to_csv(
    output_path,
    index=False
)

print("\nSaved:", output_path)
print("Final shape:", stock_prices_clean.shape)
```
