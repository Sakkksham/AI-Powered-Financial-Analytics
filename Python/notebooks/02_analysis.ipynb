```python
def parse_metric(value):
    """
    Convert values such as:
    '10 Years: 21%'
    '5 Years: 24%'
    'TTM: 43%'
    'Last Year: 12%'

    into:
    period, numeric_value
    """

    if pd.isna(value):
        return pd.Series([np.nan, np.nan])

    value = str(value).strip()

    if ":" not in value:
        return pd.Series([np.nan, np.nan])

    period, percentage = value.split(":", 1)

    period = period.strip()
    percentage = percentage.strip().replace("%", "")

    try:
        numeric_value = float(percentage)
    except ValueError:
        numeric_value = np.nan

    return pd.Series([period, numeric_value])
```

```python
def parse_metric(value):
    if pd.isna(value):
        return pd.Series([np.nan, np.nan])

    value = str(value).strip()

    # Handle both "10 Years: 21%" and "5 Years 14%"
    if ":" in value:
        period, percentage = value.split(":", 1)
    else:
        # Split at the last space before the percentage
        parts = value.rsplit(" ", 1)

        if len(parts) != 2:
            return pd.Series([np.nan, np.nan])

        period, percentage = parts

    period = period.strip()
    percentage = percentage.strip().replace("%", "")

    try:
        numeric_value = float(percentage)
    except ValueError:
        numeric_value = np.nan

    return pd.Series([period, numeric_value])
```

```python
metrics = [
    "compounded_sales_growth",
    "compounded_profit_growth",
    "stock_price_cagr",
    "roe"
]

for column in metrics:
    parsed = analysis_raw[column].apply(parse_metric)

    analysis_raw[column + "_period"] = parsed[0]
    analysis_raw[column + "_value"] = parsed[1]
```

```python
import pandas as pd
from pathlib import Path

cleaned = Path(
    r"C:\Users\KIIT\OneDrive\Desktop\Financial Analytics Project\Cleaned Data"
)

files = [
    "balance_sheet_clean.csv",
    "nifty100_analysis_clean.csv",
    "profit_loss_clean.csv",
    "cash_flow_clean.csv",
    "ratios_clean.csv",
    "stock_prices_clean.csv",
]

all_companies = set()

for file in files:
    path = cleaned / file

    if path.exists():
        temp = pd.read_csv(path)

        if "company_id" in temp.columns:
            ids = set(temp["company_id"].dropna().astype(str).str.strip())
            all_companies.update(ids)
            print(file, "→", len(ids), "companies")

print("\nTOTAL UNIQUE COMPANIES:", len(all_companies))
print(sorted(all_companies))
```

```python
master_ids = set(
    companies["id"].astype(str).str.strip()
)

all_ids = set(all_companies)

extra_ids = sorted(all_ids - master_ids)

print("Extra company IDs:", len(extra_ids))
print(extra_ids)
```

```python
for company in extra_ids:
    print("\n", company)

    for file in [
        "balance_sheet_clean.csv",
        "nifty100_analysis_clean.csv",
        "cash_flow_clean.csv"
    ]:
        path = cleaned / file

        if path.exists():
            temp = pd.read_csv(path)

            if "company_id" in temp.columns:
                count = (temp["company_id"].astype(str).str.strip() == company).sum()

                if count > 0:
                    print("  ", file, "→", count, "rows")
```
