```python
from pathlib import Path
import pandas as pd

RAW_DATA = Path(
    r"C:\Users\KIIT\OneDrive\Desktop\Financial Analytics Project\Raw Data"
)

file_path = RAW_DATA / "peer_groups.xlsx"

peer_groups = pd.read_excel(file_path)

print("Shape:", peer_groups.shape)
print("\nColumns:")
print(peer_groups.columns.tolist())

peer_groups.head()
```

```python
print("Shape:", peer_groups.shape)

print("\nData types:")
print(peer_groups.dtypes)

print("\nMissing values:")
print(peer_groups.isna().sum())

print("\nDuplicate IDs:",
      peer_groups["id"].duplicated().sum())

print("\nDuplicate company IDs:",
      peer_groups["company_id"].duplicated().sum())

print("\nPeer groups:")
print(peer_groups["peer_group_name"].value_counts())

print("\nBenchmark counts:")
print(peer_groups["is_benchmark"].value_counts())
```

```python
print("Blank peer groups:",
      peer_groups["peer_group_name"].astype(str).str.strip().eq("").sum())

print("Blank company IDs:",
      peer_groups["company_id"].astype(str).str.strip().eq("").sum())

print("Invalid benchmark values:")
print(
    peer_groups[
        ~peer_groups["is_benchmark"].isin([True, False])
    ]
)
```

```python
company_group_counts = (
    peer_groups.groupby("company_id")["peer_group_name"]
    .nunique()
    .sort_values(ascending=False)
)

print(company_group_counts.head(20))
```

```python
from pathlib import Path

CLEANED_DATA = Path(
    r"C:\Users\KIIT\OneDrive\Desktop\Financial Analytics Project\Cleaned Data\peer_groups"
)

CLEANED_DATA.mkdir(parents=True, exist_ok=True)

peer_groups_clean = peer_groups.copy()

peer_groups_clean.to_csv(
    CLEANED_DATA / "peer_groups_clean.csv",
    index=False
)

print("Saved:", CLEANED_DATA / "peer_groups_clean.csv")
```
